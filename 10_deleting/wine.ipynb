### Step 1. Import the necessary libraries

```python
import pandas as pd 
import numpy as np 
import seaborn as sns 
import matplotlib.pyplot as plt 
```

### Step 2. Import the dataset from this [address](https://archive.ics.uci.edu/ml/machine-learning-databases/wine/wine.data). 

### Step 3. Assign it to a variable called wine

```python
lnk = 'https://archive.ics.uci.edu/ml/machine-learning-databases/wine/wine.data'
wine = pd.read_csv(lnk, header=None)
wine.head()
```

### Step 4. Delete the first, fourth, seventh, nineth, eleventh, thirteenth and fourteenth columns

```python
wine = wine.iloc[:, [1, 2, 4, 5, 7, 9, 11]]
wine.head()
```

### Step 5. Assign the columns as below:

The attributes are (donated by Riccardo Leardi, riclea '@' anchem.unige.it):  
1) alcohol  
2) malic_acid  
3) alcalinity_of_ash  
4) magnesium  
5) flavanoids  
6) proanthocyanins  
7) hue 

```python
wine.columns = ['alcohol', 'malic_acid', 'alcalinity_of_ash', 'magnesium', 'flavonoids', 'proanthocyanins', 'hue']
wine.head()
```

### Step 6. Set the values of the first 3 rows from alcohol as NaN

```python
wine.iloc[:3, 0] = np.nan 
wine.head()
```

### Step 7. Now set the value of the rows 3 and 4 of magnesium as NaN

```python
wine.iloc[[2, 3], 3] = np.nan 
wine.head()
```

### Step 8. Fill the value of NaN with the number 10 in alcohol and 100 in magnesium

```python
wine.alcohol = wine.alcohol.fillna(10)
wine.magnesium = wine.magnesium.fillna(100)
wine.head()
```

### Step 9. Count the number of missing values

```python
wine.isna().sum()  # no missing values 
```

### Step 10.  Create an array of 10 random numbers up until 10

```python
arr = np.random.randint(0, 10, 10)
arr 
```

### Step 11.  Use random numbers you generated as an index and assign NaN value to each of cell.

```python
wine.iloc[arr, 0] = np.nan 
wine.head(10)
```

### Step 12.  How many missing values do we have?

```python
wine.alcohol.isna().sum()
```

### Step 13. Delete the rows that contain missing values

```python
wine = wine.dropna()
wine.isna().sum()
```

### Step 14. Print only the non-null values in alcohol

```python
wine.alcohol.loc[wine.alcohol.notna()].head(10)
```

```python
wine.alcohol.loc[wine.alcohol.notnull()].head(10)
```

### Step 15.  Reset the index, so it starts with 0 again

```python
wine = wine.reset_index(drop=True)
wine.head()
```

### BONUS: Create your own question and answer it.

```python
# again my question don't fit the theme of this dataset 
# but still let's know the relationship between alcohol and magnesium content
# i'll build regplot which shows this relationship 

sns.regplot(
    data=wine, x='alcohol', y='magnesium',
    color="#C300FF", line_kws={'color': 'darkblue'}
)
plt.title('Relationship between Alcohol and Magnesium content')
plt.tight_layout()

display(wine.alcohol.corr(wine.magnesium))  # weak positive dependence
```
