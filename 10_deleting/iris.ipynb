### Step 1. Import the necessary libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt 
```

### Step 2. Import the dataset from this [address](https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data). 

### Step 3. Assign it to a variable called iris

```python
lnk = 'https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data'
iris = pd.read_csv(lnk)
iris.head()
```

### Step 4. Create columns for the dataset

```python
iris = pd.read_csv(lnk, names=['sepal_length', 'sepal_width', 'petal_length', 'petal_width', 'class'])  # better to use names parameter 
iris.head()
```

### Step 5.  Is there any missing value in the dataframe?

```python
iris.isna().sum()  # no nan values 
```

### Step 6.  Lets set the values of the rows 10 to 29 of the column 'petal_length' to NaN

```python
iris.loc[10:29, 'petal_length'] = np.nan 
```

### Step 7. Good, now lets substitute the NaN values to 1.0

```python
iris.petal_length = iris.petal_length.fillna(1)
iris.loc[10:29]
```

### Step 8. Now let's delete the column class

```python
iris = iris.drop(columns='class')
iris.head()
```

### Step 9.  Set the first 3 rows as NaN

```python
iris.iloc[0:3] = np.nan 
iris.head()
```

### Step 10.  Delete the rows that have NaN

```python
iris = iris.dropna()
iris.head()
```

### Step 11. Reset the index so it begins with 0 again

```python
iris = iris.reset_index(drop=True)
iris.head()
```

### BONUS: Create your own question and answer it.

```python
# it's not deleting but still i'll show relationship between sepal_length & sepal_width by creating scatterplot 

plt.figure(figsize=(10, 5))

plt.scatter(x=iris.sepal_length, y=iris.sepal_width)
plt.xlabel('Sepal Length')
plt.ylabel('Sepal Width')
plt.title('Relationship between Length and Width of sepal')
plt.grid(alpha=.5, ls='--')
plt.tight_layout()
```
