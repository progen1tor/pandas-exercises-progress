### Step 4. Use Pandas Datarader to read the daily time series for the Apple stock (ticker symbol AAPL) between 01/01/2015 and today, assign it to df_apple and print it.

```python
def get_data(ticker: str) -> pd.DataFrame:
    return yf.download(
        ticker,
        start=date(2015, 1, 1),
        end=date.today()
    )

df_apple = get_data('AAPL')
df_apple.head()
```

```python
# some additional work w/df 
# decide to make a func in order not to repeat code 

def cleaner(df: pd.DataFrame) -> pd.DataFrame:
    df = df.droplevel(level=1, axis=1)
    df.columns.name = None 
    return df

df_apple = cleaner(df_apple)
df_apple.head()
```

### Step 5. Add a new column "stock" to the dataframe and add the ticker symbol

```python
df_apple['stock'] = 'AAPL'
df_apple.head()
```

### Step 6. Repeat the two previous steps for a few other stocks, always creating a new dataframe: Tesla, IBM and Microsoft. (Ticker symbols TSLA, IBM and MSFT.)

```python
df_tesla = cleaner(get_data('TSLA'))
df_tesla['stock'] = 'TSLA'

df_ibm = cleaner(get_data('IBM'))
df_ibm['stock'] = 'IBM'

df_microsoft = cleaner(get_data('MSFT'))
df_microsoft['stock'] = 'MSFT'
```

### Step 7. Combine the four separate dataFrames into one combined dataFrame df that holds the information for all four stocks

```python
df = pd.concat([df_apple, df_tesla, df_ibm, df_microsoft])
df.head()
```

### Step 8. Shift the stock column into the index (making it a multi-level index consisting of the ticker symbol and the date).

```python
df = df.set_index('stock', append=True)
df.head()
```

### Step 7. Create a dataFrame called vol, with the volume values.

```python
vol = df[['Volume']]
vol.head()
```

### Step 8. Aggregate the data of volume to weekly.
Hint: Be careful to not sum data from the same week of 2015 and other years.

```python
vol['date_col'] = vol.index.get_level_values(0)
vol['year'] = vol.date_col.dt.year 
vol['week_number'] = vol.date_col.dt.isocalendar().week 

vol_pivot = pd.pivot_table(
    data=vol, 
    index=['year', 'week_number'],
    columns='stock', 
    values='Volume'
)

vol_pivot.head()
```

### Step 9. Find all the volume traded in the year of 2015

```python
vol_2015 = vol.loc[vol['year'] == 2015]

vol_pivot_2015 = pd.pivot_table(
    data=vol_2015, 
    index='year',
    columns='stock', 
    values='Volume', 
    aggfunc='sum'
)

vol_pivot_2015.head()
```
